```python
sub1 = df[['Type','like','share']]
sub1
```

```python
#row subset
sub_row = df[df['Type'] == 'Photo']
sub_row.head()
```

```python
#subsets using loc
sub_loc = df.loc[1:10, ['Type','like']]
sub_loc
```

```python
#merging two subsets
merged_data = pd.merge(sub_loc,sub1, on= "like")
merged_data
```

```python
#sorting of data
sort_df = merged_data.sort_values(by = ['share','like'], ascending = [True,False])
sort_df
```

```python
#transposing data
trans_data = merged_data.T
trans_data
```
