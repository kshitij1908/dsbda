```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Avishkar\Downloads\Heart_dataset.csv").drop('Unnamed: 0',axis =1)
df
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['ChestPain'] = le.fit_transform(df['ChestPain'])
df['Thal'] = le.fit_transform(df['Thal'])
df['AHD'] = le.fit_transform(df['AHD'])
```

```python
df1 = df.fillna(df['Ca'].mean())
df1.isna().sum()
```

```python
# seaborn - count plot
sns.countplot(x='AHD', data = df1)
plt.title("Count plot")
plt.show()
```

```python
# matplotlib - histogram

plt.hist(df1['Age'],bins = 10)
plt.title("Histogram Plot")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#seaborn box plot

sns.boxplot(x= "AHD", y="Age", data = df1)
plt.title("Box Plot")
plt.show()
```

```python
# seaborn - heatmap

plt.figure(figsize=(10,8))
sns.heatmap(df1.corr(), annot=True, cmap="coolwarm")
plt.title("Heatmap")
plt.show()
```

```python
# matplotlib scatter plot
plt.scatter(df1['MaxHR'], df1['Age'])
plt.title("Scatter plot b/w MAXHR & Age")
plt.xlabel("MAXHR")
plt.ylabel("Age")
plt.show()
```

```python
#seaborn pairplot
sns.pairplot(df[['Age','RestBP','MaxHR','AHD']], hue='AHD')
plt.show()
```

```python
sns.barplot(x='Sex', y='AHD', data=df1)
plt.title("Sex vs Heart Disease")
plt.show()
```

```python
#seaborn violin plot
sns.violinplot(x= 'Sex', y='AHD', data = df1)
plt.show()
```

```python
sns.kdeplot(data=df1, x='Chol', hue='AHD', fill=True)
plt.title("Cholesterol Density Plot")
plt.show()
```

```python
#seaborn swarmplot
sns.swarmplot(x= 'Sex',y='AHD',data = df1)
plt.show()
```

```python
sns.stripplot(x='AHD', y='Chol', data=df1, jitter=True)
plt.title("Strip Plot of Cholesterol")
plt.show()
```
